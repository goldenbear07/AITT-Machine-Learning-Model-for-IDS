# Feature Engineering

1. To know who is talking to who, we look at the Destination port/Protocol (does not really help us with machine learning training)

2. We need to understand how long is the conversation. This could be Flow duration, Flow IAT mean std, bwd IAT std, idle... (important for IDS traffic because it helps us to kind of differenciate human browsing from automated attacks, human browsing has irregular timings, automated attacks has consistent burst of communications)

3. understand how much data is exchanged. We look at total length of fwd and bwd packets, subflow fwd and bwd packets. (Benign will receive a lot of data, malicious attacks usually will just receive a small payload)

4. understand how many packets are exchanged. Here we look at Total fwd and bwd packets, subflow fwd and bwd packets. (normal traffic will be balanced fwd and bwd packets while attack traffic will be more fwd heavy)

5. from the dataset we can also see the structure of the packet (packet length mean, std, variance, average packet size, min max packet length) (Benign will show a varied sizes from different traffics, but attack traffics will usually be small consistent payloads)

6. How fast the data is sent, rate behaviour like Flow packets, flow bytes, subflow bytes. (brute force means a lot of packets sent in a short amount of time.) helps to distinguish between different attack types.

7. how the protocol is being used, SYN, RST, ACK, PSH flags. (a lot of RST spikes means failed connections)


How the features are linked?

Flow packets/s is just total packets / Flow duration
Flow bytes/s is just total bytes / Flow duration

since these 3 features, flow, no. of packets, and duration are basically telling the same thing. we just need to keep the ratio feature which is Flow Packets, that is the one we need to focus on.

last. +1 in the feature engineerings are to prevent division by zero error

```python
# Encode categorical attack labels into numerical form for correlation analysis
# This allows mathematical correlation to be computed against the target variable
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df_corr = df.copy()
df_corr['Label_Code'] = le.fit_transform(df_corr['Label'])  # Convert attack labels (e.g., Benign, DoS, Web Attack) into numeric codes

target_corr = df_corr.drop(columns=['Label']).corrwith(df_corr['Label_Code']) #Compute correlation between each numerical feature and the encoded target label

target_corr = target_corr.drop('Label_Code') #remove label_code so it wouldnt be correlated to itself

# We use abs() (absolute value) because negative correlation is just as powerful as positive
target_corr_sorted = target_corr.abs().sort_values(ascending=False)

print("Top 10 Features highly correlated with Attack:")
print(target_corr_sorted.head(10))

plt.figure(figsize=(10, 8))
# Plot top 20 features
top_20_features = target_corr_sorted.head(25).index
sns.barplot(x=target_corr_sorted[top_20_features], y=top_20_features)
plt.title("Feature Correlation with Target (Label)")
plt.xlabel("Correlation Coefficient")
plt.show()
```

```python
# Packet direction imbalance: attacker sends more than receives (inside)
EPS = 1e-6
df["packet_ratio"] = np.log1p(df["Total Fwd Packets"] / (df["Total Backward Packets"] + 1))
df["packet_ratio"].head()
```

```python
# log transform Flow Duration helps ML models handle extreme outlier values (inside)
df["log_flow_duration"] = np.log1p(df["Flow Duration"])
```

```python
df['payload_intensity'] = (df['Total Length of Fwd Packets'] / (df['log_flow_duration'] + 1))
```

```python
df["fwd_packet_rate"] = df["Total Fwd Packets"] / (df["Flow Duration"] / 1_000_000 + 1)
```

```python
#(inside)
df["small_packet_ratio"] = (df["Min Packet Length"] / (df["Packet Length Mean"] + 1))
```

```python
# Helps separate brute bursts (many small packets) vs payload-heavy sessions (NEW) (inside)
# Different from Bytes per packet because this captures payload intensity over time
df["flow_byte_density"] = df["Flow Bytes/s"] / (df["Flow Packets/s"] + EPS)
```

```python
# How “header-heavy” a forward request is. (NEW) (inside)
# more balanced header to packet length payload means normal or brute sessions. But for XSS/ SQL usually is more header heavy
df["header_payload_ratio"] = df["Fwd Header Length"] / (df["Total Length of Fwd Packets"] + 1)
```

```python
# High CV means irregular timings; low CV means stable repeated attempts (automation) (NEW) (inside)
# IAT represents the packet intervals. Packet2 time - packet1 time.
df["fwd_IAT_coeff_var"] = df["Fwd IAT Std"] / (df["Fwd IAT Mean"] + EPS)
```

```python
#(inside)
df["packet_variability"] = df["Packet Length Std"] / (df["Packet Length Mean"] + 1)
```

```python
df['packet_direction_imbalance'] = df['Total Fwd Packets'] / (df['Total Backward Packets'] + 1)
```

```python
#(inside)
df['flow_response_ratio'] = df['Total Backward Packets'] / (df['Total Fwd Packets'] + 1)
```

```python
#inside
df['Total_Packets'] = df['Total Fwd Packets'] + df['Total Backward Packets']
df['ack_response_ratio'] = df['ACK Flag Count'] / (df['Total_Packets'] + 1)
```

```python
#inside
flag_cols = [
    'FIN Flag Count',
    'SYN Flag Count',
    'RST Flag Count',
    'PSH Flag Count',
    'ACK Flag Count',
    'URG Flag Count',
    'CWE Flag Count',
    'ECE Flag Count'
]

flag_cols = [c for c in flag_cols if c in df.columns]
df['TCP_flag_sum'] = df[flag_cols].sum(axis=1)
```

```python
# Encode categorical attack labels into numerical form for correlation analysis
# This allows mathematical correlation to be computed against the target variable
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df_corr = df.copy()
df_corr['Label_Code'] = le.fit_transform(df_corr['Label'])  # Convert attack labels (e.g., Benign, DoS, Web Attack) into numeric codes

target_corr = df_corr.drop(columns=['Label']).corrwith(df_corr['Label_Code']) #Compute correlation between each numerical feature and the encoded target label

target_corr = target_corr.drop('Label_Code') #remove label_code so it wouldnt be correlated to itself

# We use abs() (absolute value) because negative correlation is just as powerful as positive
target_corr_sorted = target_corr.abs().sort_values(ascending=False)

print("Top 10 Features highly correlated with Attack:")
print(target_corr_sorted.head(10))

plt.figure(figsize=(10, 8))
# Plot top 20 features
top_20_features = target_corr_sorted.head(25).index
sns.barplot(x=target_corr_sorted[top_20_features], y=top_20_features)
plt.title("Feature Correlation with Target (Label)")
plt.xlabel("Correlation Coefficient")
plt.show()
```

```python
engineered_cols = [
    "packet_ratio",
    "log_flow_duration",
    "flow_byte_density",
    "header_payload_ratio",
    "fwd_IAT_coeff_var",
    "packet_variability",
    "flow_response_ratio",
    "ack_response_ratio",
    "payload_intensity",
    "fwd_packet_rate",
    "packet_direction_imbalance",
    "small_packet_ratio",
    "TCP_flag_sum"
]

raw_cols = [c for c in df.columns if c not in high_corr_features and c not in engineered_cols and c != 'Label' and c != 'Label_Encoded']
print(raw_cols)
final_features = engineered_cols + raw_cols
```

# Data Balancing

```python
benign_n = 20000 #set the number of benign to 20000 traffic only, so that it would not inflate results as attack counts are low.
rng = 101

benign_df = df[df["Label"] == "BENIGN"].sample(n=benign_n, random_state=rng)
attack_df = df[df["Label"] != "BENIGN"]  # keep all attacks

df_bal = pd.concat([benign_df, attack_df], axis=0).sample(frac=1, random_state=rng).reset_index(drop=True)

print("Balanced dataset label counts:")
print(df_bal["Label"].value_counts())
```

# Predictive Modelling for Logistic regression (Benign VS attack)

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
from sklearn.metrics import mean_squared_error, root_mean_squared_error
from sklearn.metrics import ConfusionMatrixDisplay
from sklearn.multiclass import OneVsRestClassifier
```

```python
df_bal["Binary_Label"] = df_bal["Label"].isin([
    "Web Attack � Brute Force",
    "Web Attack � XSS",
    "Web Attack � Sql Injection"
]).astype(int)
df_bal[["Label", "Binary_Label"]].value_counts()
```

```python
X = df_bal[final_features]
y = df_bal["Binary_Label"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=101,
    stratify=y
)
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
model = LogisticRegression(
    solver="lbfgs",
    max_iter=5000,
    n_jobs=-1,
    class_weight="balanced"
)
model.fit(X_train_scaled, y_train)
```

# Model evaluation for Logistic Regression (Benign VS Attacks)

```python
y_pred = model.predict(X_test_scaled)
print(classification_report(y_test, y_pred))
```

```python
fig, ax = plt.subplots(figsize=(6, 5))
ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    display_labels=["Benign", "Attack"],
    ax=ax
)

ax.set_title("Confusion Matrix – Logistic Regression (Benign vs Attack)")
plt.tight_layout()
plt.show()
```

```python
mse = mean_squared_error(y_test, y_pred)
rmse = root_mean_squared_error(y_test, y_pred)
rmse
```

# Model Improving

Since there are a lot of benign traffic and so little malicious traffic, we can use SMOTE to even out the amount of benign traffic against the number of malicious traffic. This way, it will be fair for the model to have a chance to learn how benign, brute force, xss and sql traffic is like. This will most likely improve our model evaluation as the model is given a chance to learn more about how an attack traffic is like. (sql traffic is extremely small with only 6 training support)

```python
X = df_bal[final_features]
y = df_bal["Label"]
```

```python
le = LabelEncoder()
y_encoded = le.fit_transform(y)
label_mapping = dict(zip(le.classes_, le.transform(le.classes_)))
label_mapping
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.3, random_state=101,stratify=y_encoded)
```

```python
from imblearn.over_sampling import SMOTE

# Apply SMOTE (Oversampling)
# This creates new synthetic attack samples ONLY in the training set
smote = SMOTE(random_state=42)
X_train_smote, y_train_smote = smote.fit_resample(X_train, y_train)

print("New Training Shape after SMOTE:", X_train_smote.shape)
print("New Class Distribution (Everything should be balanced now):")
print(pd.Series(y_train_smote).value_counts())
```

```python

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_smote)
X_test_scaled = scaler.transform(X_test)
```

```python

model = LogisticRegression(
    solver="lbfgs",
    max_iter=2000,
    class_weight="balanced",
    n_jobs=-1,
    multi_class="multinomial"
)

model.fit(X_train_scaled, y_train_smote)
```

# Model Evaluation after model improvement

# Hyper Parameter Tunning

```python
from imblearn.pipeline import Pipeline as ImbPipeline
from sklearn.model_selection import GridSearchCV, StratifiedKFold

# Multiclass label encoding
le = LabelEncoder()
y = le.fit_transform(df_bal["Label"])
X = df_bal[final_features]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=101, stratify=y
)

# pipe is a fixed sequence of steps that every data sample must go through in the same order.
pipe = ImbPipeline(steps=[
    ("smote", SMOTE(random_state=42)),
    ("scaler", StandardScaler()),
    ("lr", LogisticRegression(max_iter=5000))
])

# param_grid defines what hyperparameters GridSearch is allowed to try.
param_grid = {
    "smote__k_neighbors": [3, 5, 7],
    "lr__C": [0.01, 0.1, 1, 10], # C higher C will flag attacks more often, more aggressive
    "lr__penalty": ["l2"],        # usually use l2 with multinomial, l1 might discard useful weak signals. If our dataset is already so imbalanced we need more weak signals so that it comes together into a big signal
    "lr__solver": ["lbfgs"],
    "lr__multi_class": ["ovr", "multinomial"],
    "lr__class_weight": [None, "balanced"]
}

#to ensure stable and fair evaluation on imbalanced data.
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=101) # creates 5 stratified folds from the training data. During cross-validation, the model is trained on 4 folds and validated on the remaining fold, repeating this process 5 times so that each fold is used once for validation. In GridSearch, every hyperparameter combination is evaluated across all 5 folds.

grid = GridSearchCV(
    pipe,
    param_grid=param_grid,
    scoring="f1_macro",   # macro is key for multiclass imbalance
    cv=cv,
    n_jobs=-1,
    verbose=1
)

grid.fit(X_train, y_train)
print("Best params:", grid.best_params_)
print("Best CV score (F1-macro):", grid.best_score_)

best_model = grid.best_estimator_
pred = best_model.predict(X_test)

print("\nConfusion matrix:\n", confusion_matrix(y_test, pred))
print("\nClassification report:\n", classification_report(y_test, pred, target_names=le.classes_))
```
